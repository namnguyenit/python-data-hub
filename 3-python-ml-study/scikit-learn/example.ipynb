```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import pandas as pd


data = pd.read_csv("./scikit-learn/study_hours_and_exam_score.csv")


data["study_hours"] = data["study_hours"].clip(lower=0)

X = data[["study_hours"]].values  
y = data["exam_score"].values


model = LinearRegression()
model.fit(X, y)


y_pred = model.predict(X)
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print(f"coef={model.coef_[0]:.4f}, intercept={model.intercept_:.4f}")
print(f"mse={mse:.4f}, r2={r2:.4f}")
moi = np.asarray([[5.5]])

print(model.predict(moi))

plt.scatter(X, y, label="Data")
plt.plot(X, y_pred, color="red", label="Linear fit")
plt.xlabel("Study hours")
plt.ylabel("Exam score")
plt.legend()
plt.tight_layout()
plt.show()



```
