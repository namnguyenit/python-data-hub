```python
import math

def tinh_tong(*numbers):
    return sum(numbers)

if __name__ == "__main__":
    print(tinh_tong(1,2,3,4,5,10))
```

```python
import math
def average(*numbers):
    return sum(numbers) /len(numbers)

if __name__ == "__main__":
    print(average(1,2,3,4))
```

```python
import math

def area(a, b, c):
    p = (a + b + c) / 2
    temp = p * (p - a) * (p - b) * (p - c)
    return math.sqrt(temp)
print(area(1,2,3))
```
